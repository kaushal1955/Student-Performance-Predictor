# 🎓 Student Performance Prediction - EDA
Exploratory Data Analysis for Student Performance Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10,6)

# Load data
df = pd.read_csv('../dataset/student_data.csv')
df.head()
```

```python
df.info()
df.describe()
```

```python
# Check missing values
df.isnull().sum()

# Performance category distribution
df['PerformanceCategory'].value_counts().plot(kind='bar', color=['#667eea','#38ef7d','#f6d365','#ff4b2b'])
plt.title('Performance Category Distribution')
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(12,8))
numeric = df.select_dtypes(include=[np.number])
sns.heatmap(numeric.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

```python
# Study Hours vs Final Score
sns.scatterplot(data=df, x='StudyHours', y='FinalScore', hue='PerformanceCategory')
plt.title('Study Hours vs Final Score')
plt.show()

# Attendance vs Final Score
sns.scatterplot(data=df, x='Attendance', y='FinalScore', hue='PerformanceCategory')
plt.title('Attendance vs Final Score')
plt.show()
```

```python
# Distribution plots
fig, axes = plt.subplots(2,3, figsize=(15,10))
cols = ['StudyHours','Attendance','PreviousScore','AssignmentScore','InternalScore','FinalScore']
for i, col in enumerate(cols):
    r, c = i//3, i%3
    sns.histplot(df[col], kde=True, ax=axes[r,c])
    axes[r,c].set_title(f'{col} Distribution')
plt.tight_layout()
plt.show()
```

```python
# Box plots by category
sns.boxplot(data=df, x='PerformanceCategory', y='FinalScore')
plt.title('Final Score by Performance Category')
plt.show()

sns.boxplot(data=df, x='Extracurricular', y='FinalScore')
plt.title('Extracurricular vs Final Score')
plt.show()
```

```python
# Pairplot
sns.pairplot(df[['StudyHours','Attendance','PreviousScore','InternalScore','FinalScore','PerformanceCategory']], hue='PerformanceCategory')
plt.show()
```

## Insights
- Attendance and PreviousScore are highly correlated with FinalScore
- StudyHours shows positive correlation
- InternetUsage has slight negative correlation
- SleepHours optimal around 7-8 hrs
